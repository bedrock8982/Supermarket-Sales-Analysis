##1. Import Libraries

```python
import pandas as pd
import matplotlib.pyplot as plt
```

##2. Load Dataset

```python
df = pd.read_csv("Stores.csv")
df.head()
```

##3. Explore Dataset

```python
print("Shape of Dataset")
print(df.shape)

print("Columns")
print(df.columns)

print("Information")
print(df.info())
```

##4. Data Cleaning

```python
df.isnull().sum()
```

##5. Statistical Analysis

```python
print("Daily Average Customers:")
print(df["Daily_Customer_Count"].mean())

print("Most Customers in a day:")
print(df["Daily_Customer_Count"].max())

print("Least Customers in a day:")
print(df["Daily_Customer_Count"].min())

print("Most Item that has been available in a single day:")
print(df["Items_Available"].max())

```

##6. Visualization

```python
plt.figure(figsize=(8,5))

plt.hist(df["Store_Sales"], bins=20)

plt.title("Store Sales Distribution")
plt.xlabel("Store Sales")
plt.ylabel("Number of Stores")

plt.savefig("store_sales_distribution.png")

plt.show()
```

```python
plt.figure(figsize=(8,5))

plt.hist(df["Daily_Customer_Count"], bins=20)

plt.title("Daily Customer Count Distribution")
plt.xlabel("Daily Customer Count")
plt.ylabel("Number of Stores")

plt.savefig("customer_distribution.png")

plt.show()
```

```python
plt.figure(figsize=(8,5))

plt.scatter(df["Store_Area"], df["Store_Sales"])

plt.title("Store Area vs Store Sales")
plt.xlabel("Store Area")
plt.ylabel("Store Sales")

plt.savefig("store_area_vs_sales.png")

plt.show()
```

```python
plt.figure(figsize=(8,5))

plt.scatter(df["Daily_Customer_Count"], df["Store_Sales"])

plt.title("Daily Customers vs Store Sales")
plt.xlabel("Daily Customer Count")
plt.ylabel("Store Sales")

plt.savefig("customers_vs_sales.png")

plt.show()
```

```python
plt.figure(figsize=(8,5))

plt.scatter(df["Items_Available"], df["Store_Sales"])

plt.title("Items Available vs Store Sales")
plt.xlabel("Items Available")
plt.ylabel("Store Sales")

plt.savefig("items_vs_sales.png")

plt.show()
```
